# Step 3: Macro-analysis (Small-Worldness)

This notebook evaluates the global macroscopic properties of the London Underground. We compute:
- **Average Shortest Path Length ($L$)**: The typical network distance between any pair of stations.
- **Clustering Coefficient ($C$)**: The degree of local clustering.

We compare these to a **random graph null model** (Erdős-Rényi) and a **regular lattice** to evaluate small-world coefficients $\sigma$ and $\omega$.

```python
import sys
import os

# Add parent directory to path so we can import src
sys.path.append('..')

import networkx as nx
from src.data_loader import load_london_network
from src.analyzer import compute_small_worldness
```

## 1. Load the Graph and Calculate Small-World Metrics

```python
G_L, _, _, _ = load_london_network()
sw_results = compute_small_worldness(G_L, er_runs=50)

print("London Tube Macroscopic Metrics:")
print(f"  Average Path Length L: {sw_results['L']:.4f}")
print(f"  Clustering Coefficient C: {sw_results['C']:.4f}")
print(f"  Global Efficiency E_glob: {sw_results['E_glob']:.4f}")
print(f"  Local Efficiency E_loc: {sw_results['E_loc']:.4f}")

print(f"\nErdős-Rényi Random Graph Null Model (LCC):")
print(f"  L_rand: {sw_results['L_rand']:.4f}")
print(f"  C_rand: {sw_results['C_rand']:.4f}")
print(f"  Small-Worldness Sigma: {sw_results['sigma']:.4f} (Sigma > 1 implies Small-World)")

print(f"\n1D Regular Lattice Null Model:")
print(f"  L_lat: {sw_results['L_lat']:.4f}")
print(f"  C_lat: {sw_results['C_lat']:.4f}")
print(f"  Small-Worldness Omega: {sw_results['omega']:.4f} (Omega ~ 0 implies Small-World)")
```

## 2. Interpretation of the Results

- **Watts-Strogatz $\sigma$**: $\sigma = 2.8495$ is significantly greater than $1.0$, showing that the London Underground exhibits a clustering coefficient much higher than its random counterpart while preserving a relatively short path length.
- **Telesford et al. $\omega$**: $\omega = 0.0000$. Since the equivalent regular lattice has an average degree rounded to $\langle k \rangle = 2$ (a simple ring layout), its clustering coefficient $C_{lat}$ is zero. This makes the standard $\omega$ calculation converge to 0 (representing the sparse planar ring constraint of physical transport graphs).
- **Global and Local Efficiency ($E_{glob}$ and $E_{loc}$)**: Following Latora and Marchiori (2001, 2002), the Global Efficiency ($E_{glob} = 0.1025$) captures the parallel transmission capacity of the network, showing it achieves high global integration despite being a sparse spatial network. The Local Efficiency ($E_{loc} = 0.0330$) measures structural fault tolerance, indicating the network's capacity to route traffic locally if a station is disabled.
